```python
import pandas as pd
import numpy as np

key = ['pre_fmt_id', 'post_fmt_id']

# ---------------------------------------------------------------
# triad_id: default is post_pre (unchanged). The dataset is appended
# only where the same pre/post pair occurs in more than one dataset
# AND the dataset is one of the known exceptions.
# ---------------------------------------------------------------
EXCEPTION_DATASETS = {'CM_IaniroFMT2', 'CM_IaniroFMT3'}

sgb_details['triad_id'] = (
    sgb_details['post_fmt_id'].astype(str) + '_' +
    sgb_details['pre_fmt_id'].astype(str)
)

n_ds = sgb_details.groupby(key)['dataset'].transform('nunique')
is_exception = (n_ds > 1) & sgb_details['dataset'].isin(EXCEPTION_DATASETS)

sgb_details.loc[is_exception, 'triad_id'] = (
    sgb_details.loc[is_exception, 'triad_id'] + '_' +
    sgb_details.loc[is_exception, 'dataset'].astype(str)
)

print("Exception rows (expect the 2 Ianiro pre/post pairs x 2 datasets):")
print(sgb_details.loc[is_exception, key + ['dataset', 'triad_id']].drop_duplicates())

# ---------------------------------------------------------------
# Diagnostics
# ---------------------------------------------------------------
print("\nRecalculating DoSER...")
print("Total rows in sgb_details:", len(sgb_details))
print("Unique triad_ids:", sgb_details['triad_id'].nunique())
print("NaN triad_ids:", sgb_details['triad_id'].isna().sum())

group_sizes = sgb_details.groupby('triad_id').size()
print("\nGroup size distribution:")
print(group_sizes.value_counts().sort_index())


def split_sgbs(s):
    """Parse a comma-separated SGB string into a set (empty set if missing)."""
    if pd.notna(s) and s != '':
        return set(s.split(','))
    return set()


# ---------------------------------------------------------------
# Aggregate to one row per triad
# ---------------------------------------------------------------
# ---------------------------------------------------------------
# DoSER denominator mode
#   'median': median of per-donor eligible SGB counts (original behaviour)
#   'union' : eligible SGBs in the union across donors (matches the numerator
#             exactly; guarantees DoSER <= 1 for multi-donor triads)
# ---------------------------------------------------------------
DENOMINATOR_MODE = 'median'
assert DENOMINATOR_MODE in {'median', 'union'}

# ---------------------------------------------------------------
# Aggregate to one row per triad
# ---------------------------------------------------------------
results = []
for triad_id, group in sgb_details.groupby('triad_id'):

    all_donor_sgbs = set()
    unique_shared_donor_post = set()
    unique_shared_donor_pre = set()
    baseline_both = set()            # SGBs shared donor-pre AND donor-post (same donor)
    per_donor_counts = []            # raw per-donor SGB counts
    per_donor_eligible_counts = []   # per-donor counts after dropping baseline_both

    for _, r in group.iterrows():
        d    = split_sgbs(r['donor_sgbs'])
        pre  = split_sgbs(r['shared_donor_pre_sgbs'])
        post = split_sgbs(r['shared_donor_post_sgbs'])

        # Baseline-persistent for THIS donor: strain shared with recipient both
        # pre- and post-FMT, so it cannot be attributed to the transplant.
        # SGBs shared donor-pre only (not post) are NOT dropped.
        both = pre & post

        all_donor_sgbs.update(d)
        unique_shared_donor_post.update(post)
        unique_shared_donor_pre.update(pre)
        baseline_both.update(both)

        per_donor_counts.append(len(d))
        per_donor_eligible_counts.append(len(d - both))

    n_unique_donor_sgbs = len(all_donor_sgbs)
    median_donor_sgbs = np.median(per_donor_counts) if per_donor_counts else np.nan

    # Numerator: SGBs shared donor-post, excluding baseline-persistent ones
    unique_shared_donor_post_only = unique_shared_donor_post - baseline_both
    doser_numerator = len(unique_shared_donor_post_only)

    # Denominator: donor SGBs, excluding the same baseline-persistent ones
    if DENOMINATOR_MODE == 'median':
        doser_denominator = (np.median(per_donor_eligible_counts)
                             if per_donor_eligible_counts else np.nan)
    else:
        doser_denominator = len(all_donor_sgbs - baseline_both)

    # NaN if nothing is left to measure after dropping baseline-persistent SGBs
    if pd.notna(doser_denominator) and doser_denominator > 0:
        doser = doser_numerator / doser_denominator
    else:
        doser = np.nan

    results.append({
        'triad_id': triad_id,
        'dataset': group['dataset'].iloc[0],
        'display_dataset': group['display_dataset'].iloc[0],
        'donor_id': ','.join(sorted(group['donor_id'].dropna().astype(str))),
        'pre_fmt_id': group['pre_fmt_id'].iloc[0],
        'post_fmt_id': group['post_fmt_id'].iloc[0],
        'post_timepoint': group['post_timepoint'].iloc[0],
        'original_donor_string': group['original_donor_string'].iloc[0],

        'donor_sgbs': ','.join(sorted(all_donor_sgbs)),
        'n_donor_sgbs': n_unique_donor_sgbs,      # raw union (reference only)
        'median_donor_sgbs': median_donor_sgbs,   # raw median (reference only)

        'pre_sgbs': group['pre_sgbs'].iloc[0],
        'post_sgbs': group['post_sgbs'].iloc[0],
        'n_pre_sgbs': group['n_pre_sgbs'].iloc[0],
        'n_post_sgbs': group['n_post_sgbs'].iloc[0],

        'shared_donor_post_sgbs': ','.join(sorted(unique_shared_donor_post)),
        'shared_donor_pre_sgbs': ','.join(sorted(unique_shared_donor_pre)),
        'n_shared_donor_post': len(unique_shared_donor_post),
        'n_shared_donor_pre': len(unique_shared_donor_pre),

        'baseline_both_sgbs': ','.join(sorted(baseline_both)),
        'n_baseline_both': len(baseline_both),

        'shared_donor_post_only_sgbs': ','.join(sorted(unique_shared_donor_post_only)),
        'n_shared_donor_post_only': doser_numerator,   # DoSER numerator

        'doser_denominator': doser_denominator,        # actual DoSER denominator

        'shared_post_pre_sgbs': group['shared_post_pre_sgbs'].iloc[0],
        'n_shared_post_pre': group['n_shared_post_pre'].iloc[0],

        'doser': doser,
    })

final_columns = [
    'triad_id', 'dataset', 'display_dataset', 'donor_id',
    'pre_fmt_id', 'post_fmt_id', 'post_timepoint', 'original_donor_string',
    'donor_sgbs', 'pre_sgbs', 'post_sgbs',
    'shared_donor_post_sgbs', 'shared_donor_pre_sgbs', 'shared_post_pre_sgbs',
    'baseline_both_sgbs',
    'n_donor_sgbs', 'median_donor_sgbs', 'n_pre_sgbs', 'n_post_sgbs',
    'n_shared_donor_post', 'n_shared_donor_pre', 'n_shared_post_pre',
    'n_baseline_both', 'n_shared_donor_post_only', 'doser_denominator',
    'doser',
]

aggregated = pd.DataFrame(results)[final_columns]

# ---------------------------------------------------------------
# Checks
# ---------------------------------------------------------------
print(f"\nRecalculated DoSER for {len(aggregated)} unique triads "
      f"(denominator mode: {DENOMINATOR_MODE})")
assert len(aggregated) == 652, f"Expected 656 triads, got {len(aggregated)}"

multi_donor = aggregated[aggregated['original_donor_string'].str.contains(',', na=False)]
if len(multi_donor) > 0:
    print(f"\nMulti-donor triads: {len(multi_donor)}")

# Numerator is a set difference, so DoSER cannot be negative
assert not (aggregated['doser'] < 0).any(), "Negative DoSER should be impossible"

invalid_doser = aggregated[aggregated['doser'] > 1]
if len(invalid_doser) > 0:
    print(f"\nWARNING: Found {len(invalid_doser)} triads with DoSER > 1")
    print(invalid_doser[['triad_id', 'n_donor_sgbs', 'n_baseline_both',
                         'n_shared_donor_post_only', 'doser_denominator',
                         'doser']].to_string(index=False))
else:
    print("\n✓ All DoSER values are in [0, 1]")

n_nan = aggregated['doser'].isna().sum()
if n_nan > 0:
    print(f"\nNote: {n_nan} triads with DoSER = NaN (empty denominator)")

print(f"Triads with >=1 baseline-persistent SGB dropped: "
      f"{(aggregated['n_baseline_both'] > 0).sum()}")

print("\nDoSER Statistics:")
print(f"  Range:  {aggregated['doser'].min():.4f} to {aggregated['doser'].max():.4f}")
print(f"  Mean:   {aggregated['doser'].mean():.4f}")
print(f"  Median: {aggregated['doser'].median():.4f}")

# aggregated.to_csv('doser_results_aggregated_090226.csv', index=False)

print("\nFirst few rows:")
print(aggregated.head())

doser_results = aggregated
```

```python
def intersect_sgb_sets(sgbs_a, sgbs_b):
    if pd.isna(sgbs_a) or sgbs_a == '' or pd.isna(sgbs_b) or sgbs_b == '':
        return np.nan
    set_a = set(sgbs_a.split(','))
    set_b = set(sgbs_b.split(','))
    return len(set_a & set_b)

# Common SGB counts (denominator)
cleaned_doser_results['n_common_donor_post'] = doser_results.apply(
    lambda r: intersect_sgb_sets(r['donor_sgbs'], r['post_sgbs']), axis=1)

cleaned_doser_results['n_common_donor_pre'] = doser_results.apply(
    lambda r: intersect_sgb_sets(r['donor_sgbs'], r['pre_sgbs']), axis=1)

cleaned_doser_results['n_common_post_pre'] = doser_results.apply(
    lambda r: intersect_sgb_sets(r['post_sgbs'], r['pre_sgbs']), axis=1)

# Strain sharing rates
cleaned_doser_results['ssr_donor_post'] = cleaned_doser_results['n_shared_donor_post'] / cleaned_doser_results['n_common_donor_post']
cleaned_doser_results['ssr_donor_pre']  = cleaned_doser_results['n_shared_donor_pre']  / cleaned_doser_results['n_common_donor_pre']
cleaned_doser_results['ssr_post_pre']   = cleaned_doser_results['n_shared_post_pre']   / cleaned_doser_results['n_common_post_pre']

print("Strain Sharing Rate Statistics:")
for col in ['ssr_donor_post', 'ssr_donor_pre', 'ssr_post_pre']:
    print(f"\n{col}:")
    print(f"  Range:  {cleaned_doser_results[col].min():.4f} to {cleaned_doser_results[col].max():.4f}")
    print(f"  Mean:   {cleaned_doser_results[col].mean():.4f}")
    print(f"  Median: {cleaned_doser_results[col].median():.4f}")
    print(f"  NaN:    {cleaned_doser_results[col].isna().sum()}")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import spearmanr, pearsonr, entropy
from sklearn.utils import shuffle
from adjustText import adjust_text

plt.rcParams['svg.fonttype'] = 'none'

# ── Shared helpers (reused by beta diversity chunk too) ───────────────────────
def format_p(p):
    return "p < 0.001" if p < 0.001 else f"p = {p:.3f}"

def sig_stars(p):
    return '***' if p < 0.001 else '**' if p < 0.01 else '*' if p < 0.05 else 'ns'

def compute_stats(df, x_col, y_col='doser', n_permutations=1000):
    """Spearman, Pearson, and permutation p-value for a pair of columns."""
    subset = df[[x_col, y_col]].dropna()
    if len(subset) < 5:
        return dict(n=len(subset), spearman_r=np.nan, spearman_p=np.nan,
                    pearson_r=np.nan, pearson_p=np.nan, perm_p=np.nan)
    x, y = subset[x_col].values, subset[y_col].values
    sp_r, sp_p = spearmanr(x, y)
    pe_r, pe_p = pearsonr(x, y)
    perm_rs = np.array([spearmanr(shuffle(x), y)[0] for _ in range(n_permutations)])
    perm_p = np.mean(np.abs(perm_rs) >= np.abs(sp_r))
    return dict(n=len(subset), spearman_r=sp_r, spearman_p=sp_p,
                pearson_r=pe_r, pearson_p=pe_p, perm_p=perm_p)

# ── 0. Build triad_diversity from scratch ──────────────────────────────────
# 0a. Filter MetaPhlAn to SGB-level rows, coerce numeric
sgb_profiles = metaphlan_profiles[
    metaphlan_profiles.index.str.contains('t__SGB', na=False)
].copy()
sgb_profiles = sgb_profiles.apply(pd.to_numeric, errors='coerce').fillna(0)
print(f"SGB-level rows: {sgb_profiles.shape[0]}, samples: {sgb_profiles.shape[1]}")

# 0b. Per-sample Shannon diversity and richness
def shannon_diversity(series):
    vals = series[series > 0].values
    if len(vals) == 0:
        return np.nan
    props = vals / vals.sum()
    return entropy(props, base=2)

def richness(series):
    return (series > 0).sum()

sample_diversity = pd.DataFrame({
    'shannon': sgb_profiles.apply(shannon_diversity, axis=0),
    'richness': sgb_profiles.apply(richness, axis=0)
})
sample_diversity.index.name = 'sample_id'

# 0c. Attach pre/post diversity metrics to each triad
def get_diversity(sample_id, metric):
    if pd.isna(sample_id) or sample_id not in sample_diversity.index:
        return np.nan
    return sample_diversity.loc[sample_id, metric]

triad_diversity = cleaned_doser_results.copy()

for metric in ['shannon', 'richness']:
    triad_diversity[f'pre_{metric}']  = triad_diversity['pre_fmt_id'].apply(lambda s, m=metric: get_diversity(s, m))
    triad_diversity[f'post_{metric}'] = triad_diversity['post_fmt_id'].apply(lambda s, m=metric: get_diversity(s, m))

triad_diversity['indication'] = triad_diversity['display_dataset'].map(indication).fillna('Unknown')

print(f"Triads with pre Shannon:  {triad_diversity['pre_shannon'].notna().sum()}")
print(f"Triads with post Shannon: {triad_diversity['post_shannon'].notna().sum()}")

# 0d. Attach donor Shannon + richness (storage only — no stats/plots here;
#     excludes multi-donor datasets)
MULTI_DONOR_DATASETS = {
    'NgS_2021', 'ShekarrizS_2025', 'CM_MORELIA', 'LeongK_2020',
    'KousgaardS_2024', 'QuraishiMN_2026'
}

DONOR_ID_COL = 'donor_id'  # <- set this to whatever column your beta-diversity chunk uses for donor sample id

multi_donor_mask = triad_diversity['display_dataset'].isin(MULTI_DONOR_DATASETS)

for metric in ['shannon', 'richness']:
    col = f'donor_{metric}'
    triad_diversity[col] = triad_diversity[DONOR_ID_COL].apply(
        lambda s, m=metric: get_diversity(s, m)
    )
    triad_diversity.loc[multi_donor_mask, col] = np.nan

print(f"Triads with donor Shannon:  {triad_diversity['donor_shannon'].notna().sum()}")
print(f"Triads with donor richness: {triad_diversity['donor_richness'].notna().sum()} "
      f"(excluded {multi_donor_mask.sum()} rows from multi-donor datasets)")

# ── Metrics: pre/post Shannon + pre/post richness only ────────────────────────
alpha_metrics = [
    ('pre_shannon',    'median_pre_shannon',    'Pre-FMT Shannon Diversity'),
    ('post_shannon',   'median_post_shannon',   'Post-FMT Shannon Diversity'),
    ('pre_richness',   'median_pre_richness',   'Pre-FMT Richness'),
    ('post_richness',  'median_post_richness',  'Post-FMT Richness'),
]

# ── Dataset medians ─────────────────────────────────────────────────────────
dataset_medians = triad_diversity.groupby('display_dataset').agg(
    median_doser          = ('doser',         'median'),
    median_pre_shannon    = ('pre_shannon',   'median'),
    median_post_shannon   = ('post_shannon',  'median'),
    median_pre_richness   = ('pre_richness',  'median'),
    median_post_richness  = ('post_richness', 'median'),
    n_triads               = ('doser',        'count')
).reset_index()
dataset_medians['indication'] = dataset_medians['display_dataset'].map(indication).fillna('Unknown')

# ── Precompute Spearman, triad-level and dataset-level (no FDR) ───────────────
alpha_triad_stats = pd.DataFrame(
    [{'metric': label, **compute_stats(triad_diversity, col)} for col, _, label in alpha_metrics]
).set_index('metric')

alpha_median_stats = pd.DataFrame(
    [{'metric': label, **compute_stats(dataset_medians, mcol, y_col='median_doser')}
     for _, mcol, label in alpha_metrics]
).set_index('metric')

print("\nTriad-level alpha diversity stats:")
print(alpha_triad_stats[['n','spearman_r','spearman_p']].to_string())
print("\nDataset-level (median) alpha diversity stats:")
print(alpha_median_stats[['n','spearman_r','spearman_p']].to_string())

# ── Plots ───────────────────────────────────────────────────────────────────
for triad_col, median_col, metric_label in alpha_metrics:

    save_svg = (triad_col == 'pre_richness')  # only save pre-FMT richness plots

    # ── Triad-level ──────────────────────────────────────────────────────────
    fig, ax = plt.subplots(figsize=(8, 6))
    fig.suptitle('Triad Level', fontsize=13)

    subset_triad = triad_diversity[['indication', 'doser', triad_col]].dropna()
    row = alpha_triad_stats.loc[metric_label]

    for ind, grp in subset_triad.groupby('indication'):
        ax.scatter(grp[triad_col], grp['doser'],
                   color=indication_colors.get(ind, '#AAAAAA'),
                   alpha=0.6, s=35, label=ind, edgecolors='none')

    if len(subset_triad) > 2:
        m, b = np.polyfit(subset_triad[triad_col], subset_triad['doser'], 1)
        x_line = np.linspace(subset_triad[triad_col].min(), subset_triad[triad_col].max(), 100)
        ax.plot(x_line, m * x_line + b, 'k--', linewidth=1.5, alpha=0.7)

    sig = sig_stars(row['spearman_p']) if pd.notna(row['spearman_p']) else 'ns'
    ax.text(0.05, 0.95,
            f"ρ = {row['spearman_r']:.2f}  {sig}\n{format_p(row['spearman_p'])}",
            transform=ax.transAxes, fontsize=10, va='top',
            bbox=dict(boxstyle='round,pad=0.35', facecolor='white', edgecolor='grey', alpha=0.85))

    ax.axhline(0, color='grey', linewidth=0.8, linestyle=':')
    ax.set_xlabel(metric_label, fontsize=11)
    ax.set_ylabel('DoSER', fontsize=11)

    indications_present = subset_triad['indication'].unique()
    handles = [plt.Line2D([0], [0], marker='o', color='w',
                           markerfacecolor=indication_colors.get(ind, '#AAAAAA'),
                           markersize=8, label=ind)
               for ind in sorted(indications_present)]
    ax.legend(handles=handles, title='Indication', bbox_to_anchor=(1.01, 0.5),
              loc='center left', fontsize=8)

    plt.tight_layout()
    if save_svg:
        fname = f"doser_vs_{triad_col}_triads_disease.svg"
        plt.savefig(fname, bbox_inches='tight')
        print(f"Saved: {fname}")
    plt.show()
    plt.close()

    # ── Dataset-level (median) ─────────────────────────────────────────────────
    fig, ax = plt.subplots(figsize=(8, 6))
    fig.suptitle('Dataset Level', fontsize=13)

    subset_med = dataset_medians[['display_dataset', 'indication', 'median_doser', median_col, 'n_triads']].dropna()
    row_med = alpha_median_stats.loc[metric_label]

    for ind, grp in subset_med.groupby('indication'):
        ax.scatter(grp[median_col], grp['median_doser'],
                   color=indication_colors.get(ind, '#AAAAAA'),
                   s=grp['n_triads'] * 10, alpha=0.85, edgecolors='k', linewidths=0.5,
                   label=ind, zorder=3)

    texts = [ax.text(r[median_col], r['median_doser'], r['display_dataset'], fontsize=7, alpha=0.9)
             for _, r in subset_med.iterrows()]
    adjust_text(texts, ax=ax, arrowprops=dict(arrowstyle='-', color='grey', lw=0.5),
                expand=(1.5, 1.8), force_text=(0.5, 0.8))

    if len(subset_med) > 2:
        m, b = np.polyfit(subset_med[median_col], subset_med['median_doser'], 1)
        x_line = np.linspace(subset_med[median_col].min(), subset_med[median_col].max(), 100)
        ax.plot(x_line, m * x_line + b, 'k--', linewidth=1.2, alpha=0.6)

    sig_med = sig_stars(row_med['spearman_p']) if pd.notna(row_med['spearman_p']) else 'ns'
    ax.text(0.95, 0.95,
            f"ρ = {row_med['spearman_r']:.2f}  {sig_med}  (n={len(subset_med)})\n{format_p(row_med['spearman_p'])}",
            transform=ax.transAxes, fontsize=9, va='top', ha='right',
            bbox=dict(boxstyle='round,pad=0.35', facecolor='white', edgecolor='grey', alpha=0.85))

    ax.axhline(0, color='grey', linewidth=0.8, linestyle=':')
    ax.set_xlabel(f'Median {metric_label}', fontsize=11)
    ax.set_ylabel('Median DoSER', fontsize=11)

    indications_present = subset_med['indication'].unique()
    handles = [plt.Line2D([0], [0], marker='o', color='w',
                           markerfacecolor=indication_colors.get(ind, '#AAAAAA'),
                           markeredgecolor='k', markersize=8, label=ind)
               for ind in sorted(indications_present)]
    ax.legend(handles=handles, title='Indication\n(dot size ∝ n triads)',
              bbox_to_anchor=(1.01, 0.5), loc='center left', fontsize=8)

    plt.tight_layout()
    if save_svg:
        fname = f"doser_vs_{median_col}_datasets_disease.svg"
        plt.savefig(fname, bbox_inches='tight')
        print(f"Saved: {fname}")
    plt.show()
    plt.close()
```

```python
from scipy.stats import spearmanr

def donor_mean_sgbs(donor_str):
    if pd.isna(donor_str):
        return np.nan
    donors = [d.strip() for d in str(donor_str).split(',')]
    vals = []
    for d in donors:
        match = sgb_details.loc[sgb_details['donor_id'] == d, 'n_donor_sgbs']
        if not match.empty:
            vals.append(match.iloc[0])
    return np.mean(vals) if vals else np.nan

triad_diversity['donor_sgb_richness'] = triad_diversity['donor_id'].apply(donor_mean_sgbs)
print(f"Triads with donor SGBs: {triad_diversity['donor_sgb_richness'].notna().sum()}")

dataset_medians_sgb = triad_diversity.groupby('display_dataset').agg(
    median_doser          = ('doser',              'median'),
    median_donor_richness = ('donor_sgb_richness',  'median'),
    n_triads               = ('doser',              'count')
).reset_index().dropna(subset=['median_doser', 'median_donor_richness'])

colors = plt.cm.tab20.colors
datasets = triad_diversity['display_dataset'].dropna().unique()
dataset_colors = {ds: colors[i % len(colors)] for i, ds in enumerate(datasets)}

fig, axes = plt.subplots(1, 2, figsize=(13, 5))

# ── Panel 1: individual triads ────────────────────────────────────────────────
ax = axes[0]
subset_ind = triad_diversity[['display_dataset', 'doser', 'donor_sgb_richness']].dropna()

for ds in datasets:
    ds_data = subset_ind[subset_ind['display_dataset'] == ds]
    ax.scatter(ds_data['donor_sgb_richness'], ds_data['doser'],
               color=dataset_colors[ds], alpha=0.5, s=35, edgecolors='none', label=ds)

if len(subset_ind) > 2:
    m, b = np.polyfit(subset_ind['donor_sgb_richness'], subset_ind['doser'], 1)
    xline = np.linspace(0, 50, 200)
    ax.plot(xline, m * xline + b, 'k--', linewidth=1.5, alpha=0.7)
    rho, pval = spearmanr(subset_ind['donor_sgb_richness'], subset_ind['doser'])
    p_str = "p < 0.001" if pval < 0.001 else f"p = {pval:.3f}"
    sig = "***" if pval < 0.001 else "**" if pval < 0.01 else "*" if pval < 0.05 else "ns"
    ax.text(0.05, 0.95, f"ρ = {rho:.3f}  {sig}\n{p_str}\nn = {len(subset_ind)} triads",
            transform=ax.transAxes, fontsize=10, va='top',
            bbox=dict(boxstyle='round,pad=0.3', fc='white', ec='lightgray', alpha=0.85))

ax.axhline(0, color='grey', linewidth=0.8, linestyle=':')
ax.set_xlabel('Donor strain richness (n SGBs)', fontsize=11)
ax.set_ylabel('DoSER', fontsize=11)
ax.set_title('DoSER vs donor strain richness\n(individual triads)', fontsize=11)
ax.spines[['top', 'right']].set_visible(False)
ax.set_xlim(0, 50)

# ── Panel 2: per-dataset medians ──────────────────────────────────────────────
ax = axes[1]
for _, row in dataset_medians_sgb.iterrows():
    ax.scatter(row['median_donor_richness'], row['median_doser'],
               color=dataset_colors[row['display_dataset']],
               s=row['n_triads'] * 10, alpha=0.85, edgecolors='k', linewidths=0.5, zorder=3)
    ax.annotate(row['display_dataset'],
                xy=(row['median_donor_richness'], row['median_doser']),
                xytext=(5, 3), textcoords='offset points', fontsize=7, alpha=0.85)

if len(dataset_medians_sgb) > 2:
    m, b = np.polyfit(dataset_medians_sgb['median_donor_richness'], dataset_medians_sgb['median_doser'], 1)
    xline = np.linspace(0, 50, 200)
    ax.plot(xline, m * xline + b, 'k--', linewidth=1.2, alpha=0.6)
    rho, pval = spearmanr(dataset_medians_sgb['median_donor_richness'], dataset_medians_sgb['median_doser'])
    p_str = "p < 0.001" if pval < 0.001 else f"p = {pval:.3f}"
    sig = "***" if pval < 0.001 else "**" if pval < 0.01 else "*" if pval < 0.05 else "ns"
    ax.text(0.05, 0.95, f"ρ = {rho:.3f}  {sig}\n{p_str}\nn = {len(dataset_medians_sgb)} datasets",
            transform=ax.transAxes, fontsize=10, va='top',
            bbox=dict(boxstyle='round,pad=0.3', fc='white', ec='lightgray', alpha=0.85))

ax.axhline(0, color='grey', linewidth=0.8, linestyle=':')
ax.set_xlabel('Median donor strain richness (n SGBs)', fontsize=11)
ax.set_ylabel('Median DoSER', fontsize=11)
ax.set_title('Median DoSER vs median donor strain richness\n(per dataset)', fontsize=11)
ax.spines[['top', 'right']].set_visible(False)
ax.set_xlim(0, 50)

handles = [plt.Line2D([0], [0], marker='o', color='w',
                       markerfacecolor=dataset_colors[ds],
                       markeredgecolor='k', markersize=7, label=ds)
           for ds in datasets]
fig.legend(handles=handles, title='Dataset\n(panel 2: dot ∝ n triads)',
           bbox_to_anchor=(1.01, 0.5), loc='center left', fontsize=8)

plt.suptitle('DoSER vs donor strain richness', fontsize=13, y=1.02)
plt.tight_layout()
plt.savefig('doser_vs_donor_sgb_richness_050.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
def save_legend(handles, title, filename, fontsize=8, title_fontsize=9):
    fig_leg = plt.figure(figsize=(3, 2))
    ax_leg = fig_leg.add_subplot(111)
    ax_leg.axis('off')
    leg = ax_leg.legend(handles=handles, title=title,
                         loc='center', fontsize=fontsize,
                         title_fontsize=title_fontsize, frameon=True)
    fig_leg.canvas.draw()
    bbox = leg.get_window_extent().transformed(fig_leg.dpi_scale_trans.inverted())
    fig_leg.savefig(filename, dpi=300, bbox_inches=bbox.expanded(1.1, 1.1))
    plt.close(fig_leg)

save_legend(indication_handles, 'Indication', 'legend_indication.png')
save_legend(abx_handles, 'ABX use', 'legend_abx.png')
```
