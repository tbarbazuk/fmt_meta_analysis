```python
import pandas as pd

# Load MetaPhlAn profiles with multi-row header
metaphlan_profiles = pd.read_csv('/shares/CIBIO-Storage/CM/scratch/users/thomas.barbazuk/meta-strain/strain_share_newrun_June26/results/metaphlan_profiles_sgb.tsv', 
                                  sep='\t',
                                  header=[0, 1, 2])

# Extract just the sample IDs (row 1 of the header)
sample_ids = [col[1] if col[1] != 'nan' else col[2] for col in metaphlan_profiles.columns]
metaphlan_profiles.columns = sample_ids

# Set first column as index
metaphlan_profiles = metaphlan_profiles.set_index(metaphlan_profiles.columns[0])
metaphlan_profiles.index.name = 'clade_name'

print(f"MetaPhlAn Profile Info:")
print(f"Shape: {metaphlan_profiles.shape}")
print(f"\nFirst 10 sample IDs (columns):")
print(metaphlan_profiles.columns.tolist()[:10])
print(f"\nFirst 10 clades (index) - showing full names:")
print(metaphlan_profiles.index.tolist()[:10])

sgb_lookup = {}
for full_name in metaphlan_profiles.index:
    # Extract the SGB portion (everything after the last '|')
    if '|' in str(full_name) and 't__SGB' in str(full_name):
        sgb_id = str(full_name).split('|')[-1]  # e.g., "t__SGB1815"
        sgb_lookup[sgb_id] = full_name
    elif 't__SGB' in str(full_name):
        # In case there's no '|' separator
        sgb_id = str(full_name)
        sgb_lookup[sgb_id] = full_name

print(f"\nCreated SGB lookup with {len(sgb_lookup)} entries")
print("Example mappings (first 10):")
for i, (sgb_id, full_name) in enumerate(list(sgb_lookup.items())[:10]):
    print(f"  {sgb_id} -> {full_name}")
```

```python
# sample_id column in filtered_samples — adjust name if it's different
sample_id_col = "sample_id"

kept_samples = set(filtered_samples[sample_id_col])

# which profile columns actually match kept_samples?
profile_samples = set(metaphlan_profiles.columns)
matched = profile_samples & kept_samples
missing_in_profiles = kept_samples - profile_samples  # in filtered_samples but no profile column
dropped_from_profiles = profile_samples - kept_samples  # profile columns being trimmed away

print(f"Profile columns: {len(profile_samples)}")
print(f"Kept samples (filtered_samples): {len(kept_samples)}")
print(f"Matched: {len(matched)}")
print(f"In filtered_samples but missing from profiles: {len(missing_in_profiles)}")
print(f"Profile columns being dropped: {len(dropped_from_profiles)}")

# trim the profile table to only the matched columns, preserving original column order
metaphlan_profiles_trimmed = metaphlan_profiles[[c for c in metaphlan_profiles.columns if c in kept_samples]]

print(f"\nTrimmed shape: {metaphlan_profiles_trimmed.shape}")

# optional: drop clades with zero abundance across all remaining samples
nonzero_mask = (metaphlan_profiles_trimmed != 0).any(axis=1)
metaphlan_profiles_trimmed = metaphlan_profiles_trimmed[nonzero_mask]
print(f"Shape after dropping all-zero clades: {metaphlan_profiles_trimmed.shape}")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams['svg.fonttype'] = 'none'

indication_colors = {
    'CDiff':                '#c23a2e',
    'Cancer':               '#7b46aa',
    'MDRB':                 '#ce7a27',
    'GvHD':                 '#b6358b',
    'Metabolic disorder':   '#366fba',
    'IBD (UC)':             '#9e9e2e',
    'IBD (CD)':             '#309139',
    'IBS':                  '#6e4635',
    'TKI induced diarrhea': '#8c8c8c',
    'Healthy Microbiome':   '#333333',
    'Psoriatic arthritis':  '#cb4d62',
    'Tourettes':            '#30a6a2',
    'ALS':                  '#2485a8',
}

fallback_color = '#bbbbbb'
dataset_col = "dataset_group"

# ── Settings ──────────────────────────────────────────────────────────────
FIG_W_MM, FIG_H_MM = 50, 75
LABEL_FS = 5          # y-label font size (pt), your minimum
AXIS_FS = 5.5
TICK_FS = 5
LABEL_OFFSET_PT = 2   # gap between bar panel edge and label
SIDE_PAD_PT = 1       # extra margin beyond the longest label
GUIDES = False        # True = faint horizontal guide lines linking labels to bars

# Legend settings
LEG_FS = 5            # legend text + title size (pt)
LEG_NCOL = 1          # raise to 2 for a shorter legend

# ──────────────────────────────────────────────────────────────────────────
# Prepare data
# ──────────────────────────────────────────────────────────────────────────

sample_datasets = set(included_samples[dataset_col].unique())
triad_datasets = set(triads["display_dataset"].unique())

missing_from_triads = sample_datasets - triad_datasets
exception = {"DuttaguptaS_2026"}
to_exclude = missing_from_triads - exception

print("Datasets in included_samples but not in triads:")
for d in sorted(to_exclude):
    print(" -", d)

filtered_samples = included_samples[
    ~included_samples[dataset_col].isin(to_exclude)
]

sample_counts = filtered_samples[dataset_col].value_counts()
triad_counts = triads["display_dataset"].value_counts()

# Order datasets by number of triads
order = triad_counts.sort_values(ascending=False).index.tolist()

sample_counts = sample_counts.reindex(order, fill_value=0)
triad_counts = triad_counts.reindex(order, fill_value=0)

indications = [indication_map.get(d, "Unknown") for d in order]
bar_colors = [indication_colors.get(ind, fallback_color) for ind in indications]

present_inds = sorted(set(indications), key=lambda x: (x == "Unknown", x))

# Display labels only (underscores -> spaces)
display_order = [d.replace("_", " ") for d in order]

# ──────────────────────────────────────────────────────────────────────────
# Figure: 50 mm wide x 75 mm tall
# ──────────────────────────────────────────────────────────────────────────

fig, (ax1, ax2) = plt.subplots(
    1, 2,
    figsize=(FIG_W_MM / 25.4, FIG_H_MM / 25.4),
    sharey=True,
    gridspec_kw={"width_ratios": [1, 1], "wspace": 0.02}
)

y = list(range(len(order)))

# Triads
ax1.barh(y, triad_counts.values, color=bar_colors, alpha=0.80, height=0.68)
ax1.set_xlabel("Triads", fontsize=AXIS_FS, labelpad=1)

# Samples
ax2.barh(y, sample_counts.values, color=bar_colors, alpha=0.80, height=0.68)
ax2.set_xlabel("Samples", fontsize=AXIS_FS, labelpad=1)

for ax in (ax1, ax2):
    ax.tick_params(axis="x", labelsize=TICK_FS, length=1.5, width=0.5, pad=0.5)
    for s in ("top", "right", "left"):
        ax.spines[s].set_visible(False)
    if GUIDES:
        for yi in y:
            ax.axhline(yi, color="0.85", lw=0.25, zorder=0)

# ──────────────────────────────────────────────────────────────────────────
# Staggered dataset labels: even rows on the left, odd rows on the right
# ──────────────────────────────────────────────────────────────────────────

ax1.set_yticks(y)
ax1.set_yticklabels([])
ax1.tick_params(axis="y", length=0)
ax2.tick_params(axis="y", left=False, labelleft=False)

ax1.set_ylim(-0.5, len(order) - 0.5)
ax1.invert_yaxis()   # highest-count dataset at top (shared with ax2)

left_texts, right_texts = [], []
for i, label in enumerate(display_order):
    if i % 2 == 0:
        t = ax1.annotate(
            label, xy=(0, i), xycoords=ax1.get_yaxis_transform(),
            xytext=(-LABEL_OFFSET_PT, 0), textcoords="offset points",
            ha="right", va="center", fontsize=LABEL_FS
        )
        left_texts.append(t)
    else:
        t = ax2.annotate(
            label, xy=(1, i), xycoords=ax2.get_yaxis_transform(),
            xytext=(LABEL_OFFSET_PT, 0), textcoords="offset points",
            ha="left", va="center", fontsize=LABEL_FS
        )
        right_texts.append(t)

# ──────────────────────────────────────────────────────────────────────────
# Set margins from the measured label widths
# ──────────────────────────────────────────────────────────────────────────

fig.canvas.draw()
renderer = fig.canvas.get_renderer()
fig_w_px = fig.bbox.width
pad_frac = (LABEL_OFFSET_PT + SIDE_PAD_PT) / (fig.get_figwidth() * 72)

left_frac = max(t.get_window_extent(renderer).width for t in left_texts) / fig_w_px + pad_frac
right_frac = max(t.get_window_extent(renderer).width for t in right_texts) / fig_w_px + pad_frac

fig.subplots_adjust(
    left=left_frac,
    right=1 - right_frac,
    bottom=0.11,
    top=1.0
)

# Diagnostics
fig.canvas.draw()
bbox = ax1.get_position()
pitch_pt = bbox.height * fig.get_figheight() * 72 / len(order)
print(f"row pitch: {pitch_pt:.2f} pt, per side: {2 * pitch_pt:.2f} pt "
      f"(label font {LABEL_FS} pt)")
print(f"left margin: {left_frac:.3f}, right margin: {right_frac:.3f}, "
      f"bar area width: {(1 - left_frac - right_frac) * FIG_W_MM:.1f} mm")

# ──────────────────────────────────────────────────────────────────────────
# Save figure
# ──────────────────────────────────────────────────────────────────────────

fig.savefig("sample_and_triad_counts_horizontal.svg", format="svg",
            bbox_inches=None, pad_inches=0)
# fig.savefig("sample_and_triad_counts_horizontal.png", dpi=600,
#             bbox_inches=None, pad_inches=0)

plt.show()
plt.close(fig)


# ──────────────────────────────────────────────────────────────────────────
# Separate legend: 80 mm wide, max 15 mm tall, 5 pt text
# ──────────────────────────────────────────────────────────────────────────
LEG_W_MM = 80
LEG_MAX_H_MM = 15
LEG_FS = 5
LEG_NCOL = 4          # 13 entries -> 4 rows; use 5 for 3 rows (even shorter)

# Legend-only display names
legend_rename = {"Healthy Microbiome": "Healthy"}
legend_labels = [legend_rename.get(ind, ind) for ind in present_inds]

legend_fig = plt.figure(figsize=(LEG_W_MM / 25.4, 1))   # height set below
legend_ax = legend_fig.add_axes([0, 0, 1, 1])
legend_ax.axis("off")

legend_handles = [
    plt.Rectangle((0, 0), 1, 1,
                  facecolor=indication_colors.get(ind, fallback_color),
                  alpha=0.80, linewidth=0)
    for ind in present_inds
]

leg = legend_ax.legend(
    legend_handles, legend_labels,
    title="Disease Indication",
    loc="upper left",
    bbox_to_anchor=(0, 0, 1, 1),
    mode="expand",
    borderaxespad=0,
    fontsize=LEG_FS, title_fontsize=LEG_FS,
    ncol=LEG_NCOL,
    frameon=True, edgecolor="black", facecolor="white", fancybox=False,
    borderpad=0.3, labelspacing=0.2,
    handlelength=0.8, handleheight=0.7, handletextpad=0.3,
    columnspacing=0.6,
)
leg.get_frame().set_linewidth(0.5)

# Fit the figure height to the legend
legend_fig.canvas.draw()
renderer = legend_fig.canvas.get_renderer()
leg_h_in = leg.get_window_extent(renderer).height / legend_fig.dpi
leg_h_mm = (leg_h_in + 0.01) * 25.4
legend_fig.set_size_inches(LEG_W_MM / 25.4, leg_h_in + 0.01)

legend_fig.savefig("sample_and_triad_counts_legend.svg", format="svg",
                   bbox_inches=None, pad_inches=0)
print(f"legend: {LEG_W_MM} mm wide x {leg_h_mm:.1f} mm tall")
if leg_h_mm > LEG_MAX_H_MM:
    print(f"WARNING: legend is taller than {LEG_MAX_H_MM} mm, increase LEG_NCOL")
plt.close(legend_fig)
```
